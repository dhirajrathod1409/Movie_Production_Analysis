```python
from bs4 import BeautifulSoup as beut
import pandas as pd
import numpy as np
import re
import requests
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time

url = "https://www.imdb.com/list/ls050782187/?sort=list_order%2Casc&page=2"

driver = webdriver.Chrome()
driver.get(url)

# Scroll to load all movies
last_height = driver.execute_script("return document.body.scrollHeight")
while True:
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(2)  # Wait for loading
    new_height = driver.execute_script("return document.body.scrollHeight")
    if new_height == last_height:
        break
    last_height = new_height

# Extract all movies
Movie_name=[]
Year=[]
Time=[]
Type=[]
metascore=[]
Rating=[]
Imdb_votes=[]
Movie_description=[]
director=[]
actors=[]
soup = beut(driver.page_source, "html.parser")
for movie in soup.find_all("h3", class_="ipc-title__text"):
    Movie_name.append(movie.text)
for i in soup.find_all("span", class_='sc-5179a348-7 idrYgr dli-title-metadata-item'):
    regex=re.findall('\\d{4}',i.text)
    if regex:
        Year.append(regex[0])
    regex2=re.findall('\\dh\\s\\d+m|\\dh',i.text)
    if regex2:
        Time.append(regex2[0])
for movie in soup.find_all("span", class_="sc-ae9e80c5-0"):
    metascore.append(movie.text)
for movie in soup.find_all("span", class_="ipc-rating-star--rating"):
    Rating.append(movie.text)
for movie in soup.find_all("span", class_="ipc-rating-star--voteCount"):
    regex=re.findall('\\((.*?)\\)',movie.text)
    if regex:
        Imdb_votes.append(regex[0])
for movie in soup.find_all("div", class_="ipc-html-content-inner-div"):
    Movie_description.append(movie.text)
for movie in soup.find_all('div',class_='sc-7746c29e-1 jtgnlK'):
    director.append(re.findall('Director(.*?)Stars',movie.text))
    actors.append(re.findall('Stars(.*)',movie.text))
driver.quit()
```
